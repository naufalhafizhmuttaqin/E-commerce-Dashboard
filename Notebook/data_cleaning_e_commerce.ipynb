**Import Libraries**

```python
# Import required libraries
import pandas as pd
import numpy as np
```

**Display Settings**

```python
# Configure dataframe display
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
pd.options.display.float_format = "{:,.2f}".format
```

**Load Dataset**

```python
# Load raw dataset
df = pd.read_excel("/content/Raw_Dataset_Ecommerce.xlsx")

# Create working copy
df_clean = df.copy()
```

**Initial Inspection**

```python
# Preview first five rows
df_clean.head()
```

```python
# Check dataset dimensions
print(f"Rows    : {df_clean.shape[0]}")
print(f"Columns : {df_clean.shape[1]}")
```

```python
# Display dataset information
df_clean.info()
```

```python
# Generate descriptive statistics
df_clean.describe(include="all").T
```

**Check Missing Values**

```python
# Count missing values
missing_values = df_clean.isna().sum()

missing_values
```

```python
# Display only columns with missing values
missing_values[missing_values > 0]
```

**Handle Missing Values**

```python
# Remove rows with missing customer ID
df_clean.dropna(subset=["CustomerID"], inplace=True)
```

```python
# Fill missing prices using product median
df_clean["UnitPrice"] = (
    df_clean.groupby("Product")["UnitPrice"]
    .transform(lambda x: x.fillna(x.median()))
)
```

```python
# Replace missing payment method
df_clean["PaymentMethod"].fillna("Unknown", inplace=True)
```

```python
# Replace missing coupon values
df_clean["CouponCode"].fillna("NONE", inplace=True)
```

```python
# Fill missing city with Unknown
df_clean["City"].fillna("Unknown", inplace=True)
```

**Check Duplicate Records**

```python
# Count duplicate rows
df_clean.duplicated().sum()
```

```python
# Remove duplicate records
df_clean.drop_duplicates(inplace=True)
```

**Standardize Text Columns**

```python
# Remove leading and trailing spaces
text_columns = [
    "Product",
    "Category",
    "City",
    "Province",
    "PaymentMethod",
    "OrderStatus",
    "CouponCode",
    "ReferralSource",
    "ShippingAddress"
]

for column in text_columns:
    df_clean[column] = df_clean[column].astype(str).str.strip()
```

```python
# Standardize text formatting
df_clean["Product"] = df_clean["Product"].str.title()
df_clean["Category"] = df_clean["Category"].str.title()
df_clean["City"] = df_clean["City"].str.title()
df_clean["Province"] = df_clean["Province"].str.title()
df_clean["OrderStatus"] = df_clean["OrderStatus"].str.title()
df_clean["ReferralSource"] = df_clean["ReferralSource"].str.title()
```

**Standardize**

```python
# Normalize payment method values
payment_mapping = {
    "transfer bank": "Transfer Bank",
    "Transfer bank": "Transfer Bank",
    "qris": "QRIS",
    "QRIS": "QRIS",
    "e-wallet": "E-Wallet"
}

df_clean["PaymentMethod"] = (
    df_clean["PaymentMethod"]
    .replace(payment_mapping)
    .str.strip()
)
```

```python
# Correct city name variations
city_mapping = {
    "Jakrta": "Jakarta",
    "Jakarta": "Jakarta",
    "Bandng": "Bandung",
    "JAKARTA": "Jakarta",
    "BANDUNG": "Bandung",
    "SURABAYA": "Surabaya"
}

df_clean["City"] = df_clean["City"].replace(city_mapping)
```

```python
# Convert order date
df_clean["OrderDate"] = pd.to_datetime(
    df_clean["OrderDate"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)
```

```python
# Check invalid dates
df_clean["OrderDate"].isna().sum()
```

```python
# Remove invalid dates
df_clean.dropna(subset=["OrderDate"], inplace=True)
```

**Convert Data Types**

```python
# Find invalid quantity values
df_clean[df_clean["Quantity"] <= 0]
```

```python
# Remove invalid quantity
df_clean = df_clean[df_clean["Quantity"] > 0]
```

**Check Outliers**

```python
# Calculate IQR
Q1 = df_clean["UnitPrice"].quantile(0.25)
Q3 = df_clean["UnitPrice"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
df_clean.describe()
```

```python
# Display outliers
outliers = df_clean[
    (df_clean["UnitPrice"] < lower) |
    (df_clean["UnitPrice"] > upper)
]

outliers
```

```python
# Remove price outliers
df_clean = df_clean[
    (df_clean["UnitPrice"] >= lower) &
    (df_clean["UnitPrice"] <= upper)
]
```

**Recalculate Total Price**

```python
# Recalculate total price
df_clean["TotalPrice"] = (
    df_clean["Quantity"] *
    df_clean["UnitPrice"] -
    df_clean["Discount"] +
    df_clean["ShippingCost"]
)
```
